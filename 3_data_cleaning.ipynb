There are some outliers in the dataset. Most of them do not make physical sense and are likely measurement errors.

I will use a 3-sigma interval to select the most realistic data and remove measurement errors.

```python
num_sigma = 3
```

Heat 7F5937 contains incorrect data.

```python
df_combined = df_combined.loc[df_combined['HEATID'] != '7F5937']

print(f"Initial number of rows: {len(df_combined)}")

columns_to_check = ['MW', 'power_on', 'power_off', 'Calc_tap_mass', 'GAS_AMOUNT', 'O2_AMOUNT']

for col in columns_to_check:
    df_combined = df_combined.loc[
        (df_combined[col] > df_combined[col].median() - num_sigma * df_combined[col].std())
        & (df_combined[col] < df_combined[col].median() + num_sigma * df_combined[col].std())
    ]

print(f"Number of rows after outlier removal: {len(df_combined)}")
```
